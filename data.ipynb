```python
import pandas as pd 
df=pd.read_csv("food_delivery_orders_dataset.csv")
df
```

```python
print("First 5 rows:")
print(df.head())
```

```python
print("\nDataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("\nMissing values before cleaning:")
print(df.isnull().sum())
```

```python
df = df.drop(['customer_rating', 'cancellation_reason'], axis=1)
```

```python
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
df
```

```python
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].str.strip()
    
```

```python
f = df.replace(r"^\s*$", pd.NA, regex=True)
df
```

```python
df = df.dropna(how="all")
```

```python

for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):# Checks whether the current column contains numerical data (integer or float).
        df[col] = df[col].fillna(df[col].median())# Missing numerical values are replaced with the column's median.
    else:
        mode_values = df[col].mode()
        if not mode_values.empty:
            df[col] = df[col].fillna(mode_values.iloc[0])
        else:
            df[col] = df[col].fillna("Unknown")
df
```

```python
df = df.drop_duplicates()# drop the duplicates 
df
```

```python
df.to_csv("cleaned_food_delivery_orders.csv", index=False)
```

```python
df['order_status'].value_counts().plot(kind='bar',title='order status distrution')
plt.xlabel('order status')
plt.ylabel('number of orders')
plt.tight_layout()
```

Conclusion – Order Status Distribution
The bar chart shows that the majority of orders are successfully completed, with approximately 49,000 completed orders. In comparison, only a very small number of orders are cancelled.

This indicates a high order completion rate and relatively low cancellation rate. The business appears to have good order fulfilment performance.

Key Insights:
- Completed orders represent the largest category.
- Cancelled orders account for a very small proportion.
- Overall order fulfilment performance appears positive.
- Further investigation of cancelled orders could help identify potential issues.
Business Recommendation: The business should maintain its current fulfilment performance and analyse cancellation reasons to reduce cancellations further and improve customer satisfaction.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["order_total"].dropna(), bins=30, kde=True)
plt.title("Distribution of Order Total")
plt.xlabel("Order Total")
plt.tight_layout()
plt.show()
```

Conclusion – Distribution of Order Total
The histogram shows that the order total distribution is positively skewed (right-skewed). Most orders have relatively low order totals, concentrated approximately between 10 and 60. As the order total increases, the number of orders gradually decreases.

A small number of orders have very high total values, extending beyond 200, which may indicate the presence of high-value purchases or potential outliers.

Overall, most customers make low-to-medium-value purchases, while only a few place high-value orders.

Key Insights:
- The distribution is right-skewed.
- Most orders have lower total values.
- High-value orders are relatively rare.
- Potential outliers may exist.
- The data is not normally distributed.
EDA Recommendation: Use a boxplot to identify potential outliers and calculate the median and mean to understand the order total distribution more accurately.

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x="order_hour")
plt.title("Number of Orders by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

Conclusion – Number of Orders by Hour
The bar chart shows the distribution of orders throughout the day. Order activity is relatively low during the early morning hours (12 AM–5 AM) and gradually increases from 6 AM onwards.

The first major peak occurs around 12 PM (noon), indicating high customer activity during lunchtime. Order volume decreases in the afternoon and rises again, reaching its highest point at 7 PM (19:00). After 8 PM, the number of orders gradually decreases.

Key Insights:
- Highest order activity occurs at 7 PM.
- Another major peak occurs at 12 PM.
- Morning and late-night hours have fewer orders.
- Customer activity is highest during lunchtime and evening hours.
- Order demand follows a clear daily pattern.
Business Recommendation: The business should ensure sufficient staff, inventory, and delivery resources during peak hours, especially between 11 AM–2 PM and 6 PM–9 PM, to improve order fulfilment and customer satisfaction.

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    df["actual_delivery_time_minutes"].dropna(),
    bins=30, kde=True
)
plt.title("Actual Delivery Time Distribution")
plt.xlabel("Delivery Time (minutes)")
plt.tight_layout()
plt.show()
```

As delivery time increases beyond 40 minutes, the number of orders gradually decreases. The distribution is positively skewed (right-skewed), indicating that only a small number of orders take significantly longer to deliver.

Key Insights:
- Most deliveries are completed within 20–40 minutes.
- The highest delivery frequency occurs around 30–35 minutes.
- Very few orders take more than 60 minutes.
- Some deliveries take over 100 minutes, indicating potential delays.
- The distribution is right-skewed and not normally distributed.
Business Recommendation: The business should focus on reducing delivery delays by improving delivery partner allocation, route optimization, and order preparation efficiency.

Overall Insight: Most orders are delivered within a reasonable time, but a small number of delayed deliveries indicate opportunities to improve delivery performance and customer satisfaction.

Conclusion – Correlation Heatmap
The correlation heatmap illustrates the relationships between different numerical variables in the dataset. It helps identify which factors are positively or negatively correlated.

The heatmap shows a strong positive correlation between subtotal, tax amount, service fee, and order total, indicating that these values tend to increase together. Similarly, estimated delivery time and actual delivery time show a strong positive relationship.

A notable finding is the negative correlation between late delivery and customer rating, suggesting that late deliveries are associated with lower customer ratings.

Key Insights:
- Order total has a strong positive correlation with subtotal and tax amount.
- Items count is positively correlated with subtotal.
- Estimated delivery time and actual delivery time are strongly correlated.
- Delivery distance has a positive relationship with actual delivery time.
- Late delivery has a negative correlation with customer rating.
- Higher delivery delays may contribute to lower customer satisfaction.
Business Recommendations:
- Improve delivery time management and route optimization.
- Reduce late deliveries to improve customer satisfaction.
- Monitor restaurant preparation time.
- Optimize pricing and additional charges.
Overall Insight: Delivery performance plays an important role in customer satisfaction. Reducing late deliveries and improving operational efficiency can help enhance the overall customer experience.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(y=df["order_total"])
plt.title("Outliers in Order Total")
plt.tight_layout()
plt.show()
```

Conclusion – Outliers in Order Total
The boxplot shows the presence of a large number of outliers in the order_total column. Most order values are concentrated approximately between ₹15 and ₹50, with a median around ₹28.

The upper whisker extends to approximately ₹95, while several data points extend beyond this value, reaching above ₹400. These points represent unusually high-value orders compared to the majority of transactions.

Key Insights:
- Most orders have relatively low total values.
- The median order total is approximately ₹28.
- Several outliers exist above ₹95.
- The distribution is positively skewed (right-skewed).
- A few customers place significantly higher-value orders.
Business Recommendation: Investigate the high-value orders to determine whether they are genuine transactions or data errors. Avoid removing outliers without understanding their business significance.

Overall Insight: The order_total column contains significant high-value outliers, indicating that a small number of orders contribute substantially higher transaction values than typical orders.
